```python
#Complete this constructor
class vehicleNodes:
    def __init__(self, brand, vehicle_type, rent, passenger, next = None):
        self.brand=brand
        self.vehicle_type=vehicle_type
        self.rent=rent
        self.passenger=passenger
        self.next=next
```

```python
class VehicleHashTable:
  def __init__(self, size):
    self.vehicleTable = [None]*size
    self.empty_slot = {}
    for i in range(size):
      self.empty_slot[i] = False

  def create_from_vehicle_info_array(self, arr):
    for i in arr:
      self.insert_vehicle(i)

  def print_vehicle_hashtable(self):
    idx = 0
    for i in self.vehicleTable:
      print(idx, ':', end = ' ')
      head = i
      while head != None:
        print(f'(Brand: {head.brand}, Type: {head.vehicle_type}, Rent: {head.rent}, No. of Passengers: {head.passenger})', end = '---->')
        head = head.next
      print('None')
      print()
      idx += 1

  def find_empty_slot(self):
    for k,v in self.empty_slot.items():
      idx = k
      del self.empty_slot[k]
      return idx


  def helper(self,string):
    if string=="":
        return 0
    else:
        return ord(string[0])+ self.helper(string[1::])

  #Do it by yourself
  def __hash_function(self, brand):
    sum=self.helper(brand)
    index= sum%len(self.vehicleTable)
    # Do not discard the next 3 lines
    if index in self.empty_slot:
        del self.empty_slot[index]
    return index


  #Do it by yourself
  def insert_vehicle(self, vehicle):
    brand,v_type,rent,passenger=vehicle
    n=vehicleNodes(brand,v_type,rent,passenger)
    idx=self.__hash_function(brand)
    if self.vehicleTable[idx]==None:
        self.vehicleTable[idx]=n
    else:
        if self.vehicleTable[idx].brand==brand:   # Important
            phead=self.vehicleTable[idx]
            n.next=phead
            self.vehicleTable[idx]=n
        else:
            idx2=self.find_empty_slot()
            self.vehicleTable[idx2]=n
```

```python
class Node_pair:
  def __init__(self, key, value, next = None):
    self.key, self.value, self.next = key, value, next


class Hashtable:
  def __init__(self, size):
    self.ht = [None]*size


  def insert(self, s):
    if self.search_hashtable(s) == 'Found':
      print(s,'Already Inserted. Cannot reinsert.')
      print('===============================')
      return
    hashed_index = self.__hash_function(s[0])
    pair = Node_pair(s[0], s[1])
    if self.ht[hashed_index] == None:
      self.ht[hashed_index] = pair
    else:
      pair.next = self.ht[hashed_index]
      self.ht[hashed_index] = pair


  def create_from_array(self, arr):
    for i in arr:
      self.insert(i)

  def print_hashtable(self):
    idx = 0
    for i in self.ht:
      print(idx, ':', end = ' ')
      head = i
      while head != None:
        print(f'(key: {head.key}, value: {head.value})', end = '-->')
        head = head.next
      print('None')
      print()
      idx += 1


  #Do by yourself
  def __hash_function(self, s):
        if len(s)%2==0:
            string=s
        else:
          string=s+'N'
        total = 0
        i = 0
        while i < len(string):
            total += int(str(ord(string[i])) + str(ord(string[i + 1])))
            i += 2
        return total%len(self.ht)



  #Do by yourself
  def search_hashtable(self, s):
    index = self.__hash_function(s[0])
    if self.ht[index] == None:
        return 'Not Found'
    else:
        temp = self.ht[index]
        while temp != None:
            if temp.key == s[0]:
                return 'Found'
            temp = temp.next
        return 'Not Found'
```

```python
class Node:
    def __init__(self, value=None, next = None):
        self.value = value
        self.next = next

class HashTable:
    def __init__(self, length):
        n = Node()
        self.ht = [n] * length
        self.length = length

    def show(self):
        count = 0
        for i in self.ht:
            temp = i
            print(count, end=" ")
            while temp!=None:
                print (temp.value, end="-->")
                temp = temp.next
            count += 1
            print()


  #Do it by yourself
    def __hash_function(self, key):
        sum=0
        if len(key)%2==0:
            for i in range(0,len(key),2):
                sum+=ord(key[i])
        else:
            for i in range(1,len(key),2):
                sum+=ord(key[i])
        idx=sum%len(self.ht)
        return idx



  #Do it by yourself
    def insert(self, key, value):
        tup1=(key,value)
        key=Node(tup1)
        index=self.__hash_function(tup1[0])
        if self.ht[index].value==None:
            self.ht[index]=key
        else:
            if self.ht[index].next==None:
                if self.ht[index].value[1]>tup1[1]:
                    self.ht[index].next=key
                else:
                    key.next=self.ht[index]
                    self.ht[index]=key
            else:
                if tup1[1]>self.ht[index].value[1]:
                    temp=self.ht[index]
                    self.ht[index]=key
                    self.ht[index].next=temp
                else:
                    temp=self.ht[index]
                    while temp.next!=None:
                        temp2=temp.next
                        if tup1[1]<temp.value[1] and tup1[1]>temp2.value[1]:
                            temp.next=key
                            key.next=temp2
                            break
                        temp=temp.next
                    temp.next=key
```

Implement the __hash_function() and remove() methods

```python
class Node_pair:
  def __init__(self, key, value, next = None):
    self.key, self.value, self.next = key, value, next


class Hashtable:
  def __init__(self, size):
    self.ht = [None]*size


  def insert(self, s):
    hashed_index = self.__hash_function(s[0])
    pair = Node_pair(s[0], s[1])
    if self.ht[hashed_index] == None:
      self.ht[hashed_index] = pair
    else:
      pair.next = self.ht[hashed_index]
      self.ht[hashed_index] = pair


  def create_from_array(self, arr):
    for i in arr:
      self.insert(i)

  def print_hashtable(self):
    idx = 0
    for i in self.ht:
      print(idx, ':', end = ' ')
      head = i
      while head != None:
        print(f'({head.key}, {head.value})', end = '-->')
        head = head.next
      print('None')
      idx += 1


  #Do it by yourself
  def __hash_function(self, key):
    index=(key+3)%len(self.ht)
    return index


  #Do it by yourself
  def remove(self, key):
    idx=self.__hash_function(key)
    if self.ht[idx]==None:
        return
    else:
        temp=self.ht[idx]
        if temp.key==key:
            self.ht[idx]=temp.next
            temp.next=None
        else:
            while temp.next!=None:
                if temp.next.key==key:
                    removed_node=temp.next
                    temp.next=temp.next.next
                    removed_node.next=None
                    #print("removed")
                temp=temp.next
```
