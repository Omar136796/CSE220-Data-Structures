```python
#Task 01: Merge Lineup
import numpy as np
def mergeLineup(pokemon_1, pokemon_2):
    result=np.array([0]*len(pokemon_1))
    j=len(pokemon_2)-1
    for o in range(len(pokemon_1)):
        if pokemon_1[o]==None:
            pokemon_1[o]=0
        if pokemon_2[o]==None:
            pokemon_2[o]=0
    for i in range(len(pokemon_1)):
        result[i]=pokemon_1[i]+pokemon_2[j]
        j-=1
    return result



print("///  Task 01: Merge Lineup  ///")
pokemon_1 = np.array([12, 3, 25, 1, None])
pokemon_2 = np.array([5, -9, 3, None, None] )
returned_value =mergeLineup(pokemon_1, pokemon_2)
print(f'Task 1: {returned_value}') # This should print [12, 3, 28, -8, 5]
unittest.output_test(returned_value, np.array([12, 3, 28, -8, 5]))

pokemon_1 = np.array([4, 5, -1, None, None])
pokemon_2 = np.array([2, 27, 7, 12, None])
returned_value =mergeLineup(pokemon_1, pokemon_2)
print(f'Task 1: {returned_value}') # This should print [4,17,6,27,2]
unittest.output_test(returned_value, np.array([4,17,6,27,2]))
```

```python
# Task 02: Discard Cards
import numpy as np
def remove(arr,idx):
    for i in range(idx,len(arr),1):
        if i<len(arr)-1:
            arr[i]=arr[i+1]
        else:
            arr[i]=0
    return arr

def discardCards(cards, t):
    count=1
    j=0
    while j<len(cards):
        if cards[j]==t:
            count+=1
            if count%2==0:
                remove(cards,j)
                j-=1
        j+=1
    return cards

print("///  Task 02: Discard Cards  ///")
cards = np.array([1,3,7,2,5,2,2,2,0])
returned_value = discardCards(cards, 2)
print(f'Task 2: {returned_value}') # This should print [1,3,7,5,2,2,0,0,0]
unittest.output_test(returned_value, np.array([1,3,7,5,2,2,0,0,0]))

cards = np.array([5,5,5,0,0])
returned_value = discardCards(cards, 5)
print(f'Task 2: {returned_value}') # This should print [5,0,0,0,0]
unittest.output_test(returned_value, np.array([5,0,0,0,0]))
```

```python
# Task 03: DUBER Fare Splitting
import numpy as np
def findGroups(money, fare):
    visit=np.array([False]*len(money))
    count=0
    for i in range(len(money)):
        if visit[i]==True:
            continue
        if money[i]!=fare:
            for j in range(i+1, len(money)):
                if money[i]+money[j]==fare:
                    if visit[i]==False and visit[j]==False:            # Important
                        visit[i]=True
                        visit[j]=True
                        count+=1
                        print(f"Group {count} : {money[i]}, {money[j]}")
        if money[i]==fare:
            visit[i]=True
            count+=1
            print(f"Group {count} : {money[i]}")

    if False in visit:
        for i in range(len(visit)):
            if i==0:
                print(f"Ungrouped : ",end="")
            if visit[i]==False:
                print(money[i],end=" ")



print("///  Task 03: DUBER Fare Splitting  ///")
money = np.array( [120, 100, 150, 50, 30])
fare = 150
print(f'Task 3:')
findGroups(money, fare) # This should print

# Group 1 : 120, 30
# Group 2 : 100, 50
# Group 3 : 150


money = np.array( [60, 150, 60, 30, 120, 30])
fare = 180
print(f'Task 3:')
findGroups(money, fare) # This should print

# Group 1 : 60, 120
# Group 2 : 30, 150
# Ungrouped : 30 60
```

```python
#Assignment Part-1
#Write 3 methods and driver codes for this part.
import numpy as np
def mean(a):
  x=0
  for i in a:
    x+=i
  m=x/len(a)
  return m
def std(b):
  x=0
  m=mean(b)
  for i in b:
    x+=(i-m)**2
  s=(x/(len(b)-1))**.5
  return s
def newar(c):
  m=mean(c)
  s=std(c)
  r=m-(s*1.5)
  l=m+(s*1.5)
  x=0
  for i in c:
    if r>=i or i>=l:
      x+=1

  newarray=np.array([0]*x)
  j=0
  while j<len(newarray):
    for i in c:
       if r>=i:
        newarray[j]=i
        j+=1
       elif i>=l:
        newarray[j]=i
        j+=1
  return newarray
arr = np.array([10,8,13,9,14,25,-5,20,7,7,4])
print(f"Mean: {mean(arr)}")
print(f"Standard deviation: {std(arr)}")
print(f"New array: {newar(arr)}")
```
