**Assignment Part 1: Doubly Linked List**

```python
#Assignment Part 1

class Patient:
  def __init__(self, id, name, age, bg):
        self.patient_id, self.patient_name, self.patient_age, self.patient_bloodgroup = id, name, age, bg
        self.next, self.prev = None, None
```

```python
class WRM:
    def __init__(self):
        self.head = Patient(None, None, None, None)
        self.head.next, self.head.prev = self.head, self.head
        self.temp = self.head

    def RegisterPatient(self, id, name, age, bg):
        np = Patient(id, name, age, bg)
        self.temp.next = np
        np.prev = self.temp
        np.next = self.head
        self.head.prev = np
        self.temp = np
        print("Patient registered.")

    def ServePatient(self):
        p = self.head.next
        if p != self.head:
            print(f"{p.patient_id} is being served.")
            self.head.next = p.next
            p.next.prev = self.head
        else:
            print("No patient available.")

    def CancelAll(self):
        self.__init__()
        print("All appoinments cancelled.")

    def CanDoctorGoHome(self):
        if self.head.next == self.head:
            return True
        return False

    def ShowAllPatient(self):
        p = self.head.next
        if p != self.head:
            print("Patient list: ")
            while p != self.head:
                print(p.patient_id, end = " ")
                p = p.next
            print()
        else:
            print("No patients available.")

    def ReverseTheLine(self):
        if self.head.next != self.head:
            self.head.next = self.head.prev
            cp = self.head.next

            while cp != self.head:
                tp = cp.prev
                tn = cp.next
                cp.next = tp
                cp.prev = tn
                pp = cp
                cp = tp

                if cp == self.head:
                    self.head.prev = pp
                    pp.next = self.head
        self.ShowAllPatient()

```

```python
#Write a Tester Code in this cell
wrm = WRM()

print("Choose an option...")

while True:
    print("1. RegisterPatient()\n2. ServePatient()\n3. ShowAllPatients()\n4. CanDoctorGoHome()\n5. CancelAll()\n6. ReverseTheLine()\n7. exit")
    option = int(input("Choose an option: "))
    if option == 1:
        id = int(input("Patient ID: "))
        name = input("Patient name: ")
        age = int(input("Patient age: "))
        bg = input("Patient bloodgroup: ")
        wrm.RegisterPatient(id, name, age, bg)
    elif option == 2:
        wrm.ServePatient()
    elif option == 3:
        wrm.ShowAllPatient()
    elif option == 4:
        can = wrm.CanDoctorGoHome()
        print(f"Can Doctor go home? {can}")
    elif option == 5:
        wrm.CancelAll()
    elif option == 6:
        wrm.ReverseTheLine()
    elif option == 7:
        break
if option == 7:
    print("Exited.")
```

**Assignment Part 2: Stack**

Linked List based Stack is implemented in the following cell.

```python
class Node:
  def __init__(self,elem=None,next=None):
    self.elem = elem
    self.next = next

class Stack:
  def __init__(self):
    self.__top = None

  def push(self,elem):
    nn = Node(elem,self.__top)
    self.__top = nn

  def pop(self):
    if self.__top == None:
      #print('Stack Underflow')
      return None
    e = self.__top
    self.__top = self.__top.next
    return e.elem

  def peek(self):
    if self.__top == None:
      #print('Stack Underflow')
      return None
    return self.__top.elem

  def isEmpty(self):
    return self.__top == None
```

Task 1: Diamond Count

```python
def diamond_count(stack,string):
    count=0
    for i in string:
        if i=="<":
            stack.push(i)
        elif i==">":
            p=stack.pop()
            if p!=None:
                count+=1
    return count




print('Test 01')
stack = Stack()
string = '<..><.<..>> '
returned_value = diamond_count(stack,string)
print(f'Number of Diamonds: {returned_value}') #This should print 3
unittest.output_test(returned_value, 3)
print('-----------------------------------------')


print('Test 02')
stack = Stack()
string = '<<<..<......<<<<....>'
returned_value = diamond_count(stack,string)
print(f'Number of Diamonds: {returned_value}') #This should print 1
unittest.output_test(returned_value, 1)
print('-----------------------------------------')


print('Test 03')
stack = Stack()
string = '>>><...<<..>>...>...>>>'
returned_value = diamond_count(stack,string)
print(f'Number of Diamonds: {returned_value}') #This should print 3
unittest.output_test(returned_value, 3)
print('-----------------------------------------')
```

Task 2: Tower of Blocks

```python
def remove_block(stack, n):
    st2=Stack()
    for i in range(n):
        st2.push(stack.pop())
    st2.pop()
    while st2.isEmpty()!=True:
        stack.push(st2.pop())
    return stack



print('Test 01')
st = Stack()
st.push(4)
st.push(19)
st.push(23)
st.push(17)
st.push(5)
print('Stack:')
print_stack(st)
print('------')
remove_block(st,2)
print('After Removal')
print_stack(st)
print('------')

print()
print('======================================')
print()

print('Test 02')
st = Stack()
st.push(73)
st.push(85)
st.push(15)
st.push(41)
print('Stack:')
print_stack(st)
print('------')
remove_block(st,3)
print('After Removal')
print_stack(st)
print('------')

print()
print('======================================')
print()
```

```python
def conditional_reverse(stack):
    st1=Stack()
    while stack.isEmpty()!=True:
        if stack.peek()!=st1.peek():  # driver code e top private kora tai eikhan theke access kora jabe na
            st1.push(stack.pop())
        else:
            stack.pop()
    return st1



print('Test 01')
st=Stack()
st.push(10)
st.push(10)
st.push(20)
st.push(20)
st.push(30)
st.push(10)
st.push(50)
print('Stack:')
print_stack(st)
print('------')
reversed_stack=conditional_reverse(st)
print('Conditional Reversed Stack:')
print_stack(reversed_stack) # This stack contains 50, 10, 30, 20, 10 in this order whereas top element should be 10
print('------')
```
