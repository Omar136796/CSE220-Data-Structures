```python
class BTNode:
  def __init__(self, elem):
    self.elem = elem
    self.right = None
    self.left = None
```

```python
def inorder(root):
  if root == None:
    return

  inorder(root.left)
  print(root.elem, end = ' ')
  inorder(root.right)
```

```python
def tree_construction(arr, i = 1):
  if i>=len(arr) or arr[i] == None:
    return None
  p = BTNode(arr[i])
  p.left = tree_construction(arr, 2*i)
  p.right = tree_construction(arr, 2*i+1)
  return p


root2 = tree_construction([None, 'A', 'B', 'C', 'D', 'E', 'F', 'G', 'H', None, None, None, 'I', 'J', None, 'k'])
inorder(root2)
```

```python
def convert_mirror(root):
    if root==None:
        return None
    else:
        n=BTNode(root.elem)
        n.left=convert_mirror(root.right)
        n.right=convert_mirror(root.left)
    return n

#DRIVER CODE
root = BTNode(10)
n1 = BTNode(20)
n2 = BTNode(30)
n3 = BTNode(40)
n4 = BTNode(60)

root.left = n1
root.right = n2

n1.left = n3
n1.right = n4

print('Given Tree Inorder Traversal: ', end = ' ')
inorder(root) #Given Tree Inorder Traversal:  40 20 60 10 30
print()

root2 = convert_mirror(root)
print('Mirrored Tree Inorder Traversal: ', end = ' ')
inorder(root2) #Mirrored Tree Inorder Traversal:  30 10 60 20 40
```

```python
def helper(root,level=0,dict1={}):
    if root==None:
        return dict1
    else:
        if level not in dict1.keys():
            dict1[level]=root.elem
        else:
            dict1[level]=min(root.elem,dict1[level])
    helper(root.left,level+1,dict1)
    helper(root.right,level+1,dict1)
    return dict1
def smallest_level(root):
    x=helper(root,dict1={})     # eikhane dict1={} na pathaile arekbar function call korle ager value dictionary te saved thake
    return x

#DRIVER CODE
root = tree_construction([None, 4,9,2,3,-5,None,7])
print('Given Tree Inorder Traversal: ', end = ' ')
inorder(root) #Given Tree Inorder Traversal:  3 9 -5 4 2 7
print()
print('Level Wise Smallest Value: ', end = ' ')
print(smallest_level(root)) #Level Wise Smallest Value:  {0: 4, 1: 2, 2: -5}
```

```python
def maximum(left): #rightmost node of left node founder
  if left.right==None:
        return left
  return maximum(left.right)

def last_right_parent(root,x,parent=None): # For leaf Node
    if root==None:
      return None
    if root.elem==x.elem:
      if parent!=None:
        return parent
      else:
        return x
    else:
      if x.elem<root.elem:
        return last_right_parent(root.left, x, parent)
      else:
        return last_right_parent(root.right, x,root)


def inorder_predecessor(root, x):
  if root==None:
    return None
  if x.left==None:
      return last_right_parent(root,x)
  return maximum(x.left)

root = BTNode(20)
n1 = BTNode(8)
n2 = BTNode(22)
n3 = BTNode(4)
n4 = BTNode(12)
n5 = BTNode(10)
n6 = BTNode(14)

root.left = n1
root.right = n2

n1.left = n3
n1.right = n4

n4.left = n5
n4.right = n6

print('Given Tree Inorder Traversal: ', end = ' ')
inorder(root) #Given Tree Inorder Traversal:  4 8 10 12 14 20 22
print()

x = root
print(f'Inorder predecessor of node {x.elem}: {inorder_predecessor(root, x).elem}') #Inorder predecessor of node 20: 14
```

```python
def swap_child(root, level, M):
    if root==None:
        return None
    else:
        if level==M:
            return root #return or return None dile milbe na
        else:
            n=BTNode(root.elem)
            n.left=swap_child(root.right,level+1,M)
            n.right=swap_child(root.left,level+1,M)
        return n

#Driver Code
root=BTNode('A')
#Write other nodes by yourself from the given tree of Doc File
n1 = BTNode('B')
n2 = BTNode('C')
n3 = BTNode('D')
n4 = BTNode('E')
n5 = BTNode('F')
n6 = BTNode('G')
n7 = BTNode('H')
n8 = BTNode('I')
n9 = BTNode('J')

root.left=n1
root.right=n2
n1.left=n3
n1.right=n4
n3.left=n6
n3.right=n7
n4.left=n8
n2.right=n5
n5.left=n9



print('Given Tree Inorder Traversal: ', end = ' ')
inorder(root)   #Given Tree Inorder Traversal: G D H B I E A C J F
print()

root2 = swap_child(root, 0, 2)
print('Swapped Tree Inorder Traversal: ', end = ' ')
inorder(root2)  #Swapped Tree Inorder Traversal: J F C A I E B G D H
```
